```python
missing_data=pd.DataFrame({
    "Missing_count": df.isnull().sum(),
    "Missing_percentage": (
        df.isnull().mean() * 100
    ).round(2)
})
missing_data=missing_data.sort_values(
    by="Missing_percentage",
    ascending=False
)
display(missing_data)
```

```python
df.columns=df.columns.str.strip()
```

```python
date_cols=['invoice_date','due_date','payment_date']
for col in date_cols:
    df[col]=pd.to_datetime(df[col], errors='coerce')
```

```python
text_cols = [
    "invoice_number", "customer_id",
    "customer_name", "customer_email",
    "customer_city", "customer_state",
    "customer_country", "vendor_id",
    "vendor_name", "product_id",
    "product_category", "product_name",
    "currency", "payment_method",
    "payment_status", "order_channel",
    "sales_rep"
]
for col in text_cols:
    df[col] = df[col].astype("string").str.strip()
df["customer_email"] = df["customer_email"].str.lower()
df["currency"] = df["currency"].str.upper()
for col in [
    "customer_name", "vendor_name",
    "product_name", "product_category",
    "payment_method", "payment_status",
    "order_channel", "sales_rep"
]:
    df[col] = df[col].str.title()
print("Text Cleaning Completed!")
```

```python
numeric_cols=["quantity","unit_price","discount_percent","tax_percent","subtotal","discount_amount","tax_amount","shipping_cost","total_amount"]
for col in numeric_cols:
    df[col]=pd.to_numeric(
      df[col],
      errors="coerce")
```

```python
for col in numeric_cols:
    negative_count=(df[col]<0).sum()
    print(
        col,
       "Negative values:",
    negative_count
)
```

```python
invalid_dates = df[
    df["due_date"] < df["invoice_date"]]
print(
    "Invalid Due Date Records:",
    len(invalid_dates)
)
invalid_payment_dates = df[
    df["payment_date"] < df["invoice_date"]]
print(
    "Invalid Payment Date Records:",
    len(invalid_payment_dates)
)
```

```python
plt.figure(figsize=(10, 5))
sns.boxplot(
    x=df["total_amount"].sample(
        min(5000, len(df)),
        random_state=42))
plt.title("Total Amount - Outlier Detection")
plt.xlabel("Total Amount")
plt.show()
```

```python
status = df["payment_status"].str.lower()
paid = status.eq("paid").fillna(False)
open_statuses = [
    "pending",
    "overdue",
    "open",
    "unpaid",
    "partially paid"
]
open_mask = status.isin(open_statuses)
cancelled = status.eq("cancelled")
df["days_to_pay"] = np.nan
valid_payment = (
    paid &
    df["payment_date"].notna() &
    df["invoice_date"].notna()
)
df.loc[valid_payment, "days_to_pay"] = (
    df.loc[valid_payment, "payment_date"] -
    df.loc[valid_payment, "invoice_date"]
).dt.days
df["delay_days"] = np.nan

valid_delay = (
    valid_payment &
    df["due_date"].notna()
)
df.loc[valid_delay, "delay_days"] = (
    df.loc[valid_delay, "payment_date"] -
    df.loc[valid_delay, "due_date"]
).dt.days
df["paid_late_flag"] = pd.Series(
    pd.NA,
    index=df.index,
    dtype="Int64"
)
df.loc[valid_delay, "paid_late_flag"] = (
    df.loc[valid_delay, "delay_days"] > 0
).astype(int)
df["open_paid_flag"] = pd.Series(
    pd.NA,
    index=df.index,
    dtype="string"
)
df.loc[paid, "open_paid_flag"] = "Paid"
df.loc[open_mask, "open_paid_flag"] = "Open"
df.loc[cancelled, "open_paid_flag"] = "Cancelled"
print("Feature Engineering Completed!")
print(df["open_paid_flag"].value_counts(dropna=False))
```

```python
cleaning_log = []

def add_log(issue, action, reason, rows):

    cleaning_log.append({
        "Issue": issue,
        "Action": action,
        "Reason": reason,
        "Rows_Affected": int(rows)
    })
add_log(
    "Duplicate Rows",
    "Exact duplicates removed",
    "Avoid repeated records",
    removed_rows
)
add_log(
    "Duplicate Invoice IDs",
    "Flagged for investigation",
    "Check invoice identity",
    df["invoice_id"].duplicated().sum()
)
add_log(
    "Missing Values",
    "Checked and retained",
    "Avoid deleting meaningful information",
    df.isnull().any(axis=1).sum()
)
add_log(
    "Invalid Due Dates",
    "Flagged for investigation",
    "Due date before invoice date",
    len(invalid_dates)
)
add_log(
    "Negative Amounts",
    "Flagged for investigation",
    "Check possible data errors or adjustments",
    (df[numeric_cols] < 0).any(axis=1).sum()
)
cleaning_log_df = pd.DataFrame(cleaning_log)
display(cleaning_log_df)
cleaning_log_df.to_csv(
    "cleaning_log.csv",
    index=False
)
```
